```python
import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Input
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
```

```python
df_merged = pd.read_csv('master_merged_adjusted.csv')
print(df_merged.shape)
df_merged.head()
```

```python
df_merged['CASERNE'].unique()
```

```python
dict_casernes = {}

for cas_no in df_merged['CASERNE'].unique():
  # Put each set of caserne data into its own data frame
  dict_casernes[cas_no] = df_merged[df_merged['CASERNE'] == cas_no].reset_index(drop=True).sort_values(by=['YEAR', 'MONTH'], ascending=[True, True])
```

```python
def create_regression_model(input_shape):
  model = Sequential()
  model.add(Input(shape=input_shape))
  model.add(SimpleRNN(48, activation='relu', return_sequences=True))
  model.add(SimpleRNN(48, activation='relu'))
  model.add(Dense(1))
  model.compile(optimizer=Adam(learning_rate=0.001), loss='mean_squared_error', metrics=['accuracy'])
  return model
```

```python
np.random.seed(42)
tf.random.set_seed(42)

models = {}
history = {}
results = {}

for cas_no in dict_casernes.keys():
  scaler = MinMaxScaler()
  X = dict_casernes[cas_no][[
    'YEAR',
    'MONTH_SIN',
    'MONTH_COS',
    'LAG1_INCIDENT_COUNT',
    'TEMPERATURE_MAX'
  ]]
  y = dict_casernes[cas_no]['INCIDENT_COUNT'].values

  X_scaled = scaler.fit_transform(X)
  X_scaled = X_scaled.reshape(X_scaled.shape[0], 1, X_scaled.shape[1])

  models[cas_no] = create_regression_model(X_scaled.shape[1:])
  results[cas_no] = []

  tscv = TimeSeriesSplit(n_splits=5, test_size=12)

  for train_index, test_index in tscv.split(X_scaled):
      # Split the data into training and test sets for each fold
      X_train, X_test = X_scaled[train_index], X_scaled[test_index]
      y_train, y_test = y[train_index], y[test_index]

      early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
      history[cas_no] = models[cas_no].fit(
        X_train,
        y_train,
        epochs=100,
        batch_size=12,
        validation_data=(X_test, y_test),
        callbacks=[early_stopping],
        verbose=0
      )
      # Evaluate the model on the test set
      y_pred = models[cas_no].predict(X_test)
      mse = mean_squared_error(y_test, y_pred)

      # Store the metrics
      results[cas_no].append(round(mse, 3))

  # Calculate the average of the metrics across all folds
  avg_mse = round(np.mean(results[cas_no]), 3)

  print(f'Caserne {cas_no} MSE: {results[cas_no]}')
  print(f'Caserne {cas_no} Average MSE: {avg_mse}')
```

```python
dict_means = {}
avg_split_mse = np.array([0, 0, 0, 0, 0])

for cas_no in dict_casernes.keys():
  if (cas_no != 79):
    dict_means[cas_no] = round(np.mean(results[cas_no]), 3)
    avg_split_mse = avg_split_mse + np.array(results[cas_no])

avg_split_mse = np.round(avg_split_mse / len(dict_means), 3)

min_cas = min(dict_means, key=dict_means.get)
max_cas = max(dict_means, key=dict_means.get)

print(f'Mean MSE for each test split: {avg_split_mse}')
print(f'Min Mean MSE:  Caserne {min_cas}, {dict_means[min_cas]}')
print(f'Max Mean MSE: Caserne {max_cas}, {dict_means[max_cas]}')
print(f'Overall Mean MSE: {round(np.mean(avg_split_mse), 3)}')
```
